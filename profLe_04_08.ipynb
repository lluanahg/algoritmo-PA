```python
# Crie as variáveis abaixo e mostre o valor e o tipo de cada uma usando print() e type()


nome = input("Digite nome do aluno: ")

idade = int(input("Digite idade do aluno: "))

altura = float(input("Digite altura aqui: "))
nota = float(input("Digite a nota final aqui: "))
cidade = input("Digite cidade aqui: ")

print("Informações do tipo de variável abaixo:" )


print(type(nome))
print(type(idade))
print(type(altura))
print(type(nota))
print(type(cidade))
```
