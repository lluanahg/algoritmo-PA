Escrever um programa que faz a conversão de temperatura

entrada: Temperatura em Celsius

Processamento: converter para fahrenheit
Saída: resultado.
formula: F = (C x 9/5) + 32

```python
# Escreva um programa que faz conversão de temperatura

# Entrada

Celsius = float (input("Digite a temperatura em celsius: "))

 # Processamento

F = Celsius * (9/5) + 32

# Saída

print (f"temperatura em Fahrenheit: {F}")
```
