```python
# Classificação de Idade

idade = int(input("Digite sua idade aqui: "))

if idade >= 0 and  idade <= 12:
   print("Você é uma criança")

elif  idade >= 13 and idade <= 17:
      print("Você é um adolescente")

elif idade >= 18 and idade <= 59:
     print("Você é um adulto")

elif idade >= 60:
     print("Você é idoso")

```

```python
#Positivo negativo ou zero

numero_str = input("Digite um número: ")

try:
    numero = float(numero_str.replace(',', '.'))

    if numero > 0:
       print("Esse número é positivo ")
    elif numero < 0:
         print("Esse número é negativo")
    else: # numero == 0
         print("Esse número é zero")
except ValueError:
    print("Entrada inválida. Por favor, digite um número válido.")
```

```python
#Calculadora simples

print("-------------MENU-----------")
print("")
print (" Opcão - 1 = SOMA")
print(" Opção - 2 = SUBTRAÇÃO")
print (" Opção - 3 = MULTIPLICAÇÃO")
print (" Opção - 4 = DIVISÃO")
print("")

opcao = int (input("Digite a opção: "))

# Entrada

numero1 = float (input("Digite o primeiro número: "))
numero2 = float (input("Digite o segundo número: "))

if opcao == 1:
  resultado = numero1 + numero2
  print (f"O resultado é {resultado} ")

elif opcao == 2:
     resultado = numero1 - numero2
     print (f"O resultado é {resultado} ")

elif opcao == 3:
     resultado = numero1 * numero2
     print (f"O resultado é {resultado} ")

elif opcao == 4:
    if numero2 != 0:
     resultado = numero1 / numero2
     print (f"O resultado é {resultado} ")
    else:
      print (f"Erro: divisão por zero!")

else:
   print(f"Opção inválida!")
```

```python
# Dia da semana

numero = int (input("Digite um número de 1 a 7: "))

if numero == 1:
   print("Domingo")

elif numero == 2:
   print("Segunda-feira")

elif numero == 3:
     print("terça-feira")


elif numero == 4:
     print("Quarta-feira")

elif numero == 5:
     print("Quinta-feira")


elif numero == 6:
     print("Sexta-feira")

elif numero == 7:
     print("Sábado")

else:
  print("Número inválido")

```

```python
# Desconto em loja

valor = float(input("Digite o valor da compra: R$ "))

if valor >= 100:
   desconto = 0.10

elif valor >= 50:
     desconto = 0.05

else:
  desconto = 0


valor_desconto = valor * desconto
total = valor - valor_desconto


print("Desconto aplicado:", desconto * 100,"%")

print("Valor do desconto: R$",valor_desconto)

print("Total a pagar: R$", total)
```
