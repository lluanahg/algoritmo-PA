```python
# 5 Soma, o usuário digita um número e o programa soma os números do zero até o número

numero = int(input("Digite um número inteiro e positivo aqui: "))

if numero < 0:
   print("Digite um número valido: ")

else:
    soma_acumulada = 0
    for i in range (1, numero + 1):
        soma_acumulada += i
    print (f"A soma acumulada de 1 até {numero} é: {soma_acumulada})")
```

```python
# 11 Pirâmide Numérica Simétrica

# Faça um programa que receba um número N e imprima uma pirâmide perfeitamente centralizada

# Exemplo
numero = int(input("Digite um número para a pirâmide: "))

for i in range(1, numero + 1):
    # Calculate leading spaces for centering
    espacos = " " * (numero - i)

    # Generate ascending part of numbers (e.g., 1, 12, 123)
    parte_ascendente = ""
    for j in range(1, i + 1):
        parte_ascendente += str(j)

    # Generate descending part of numbers (e.g., empty, 1, 21)
    parte_descendente = ""
    for j in range(i - 1, 0, -1):
        parte_descendente += str(j)

    # Combine and print the row
    print(f"{espacos}{parte_ascendente}{parte_descendente}")
```
