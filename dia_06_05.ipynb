```python
# 1 Contagem simples, o usuario digita um número e o programa imprime do zero até o número.

numero = int(input("Digite um número: "))

for i in range (0, numero + 1):
    print (i)
```

```python
# 2 Contagem regressiva, o usuário digita um número e o programa imprime os números até o zero

numero = int(input("Digite um número e faça sua contagem regressiva: "))

for i in  range (numero, -1, -1):
    print (i)


```

```python
# 5 Soma, o usuário digita um número e o programa soma os números do zero até o número

numero = int(input("Digite um número inteiro e positivo aqui: "))

if numero < 0:
   print("Digite um número valido: ")

else:
    soma_acumulada = 0
    for i in range (1, numero + 1):
        soma_acumulada += i
    print (f"A soma acumulada de 1 até {numero} é: {soma_acumulada})")
```

```python
# 6 Tabuada, o usuário digita um número e o programa imprime a tabuada do número

numero = int(input("Digite um número e descubra sua tabuada: "))

for i in range (1, 11):
  print(f"{numero} x {i} = {numero * i}")
```
