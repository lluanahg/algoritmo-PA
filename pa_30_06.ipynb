```python
 # Cofre digital
 # Um cofre possui um código composto por 5 números inteiros.

 # Desenvolva um programa que:

 # Leia os 5 números utilizando um laço de repetição.

 # Ao final informar:

 # 1 - Quantos números são pares.
 # 2 - Quantos números são impares
 # 3 - O maior número
 # 4 - A soma de todos os números.
 # 5 - A média dos números.
 # 6 - O menor número.
 # 7 - Qantos números são maiores que a media.
```

```python
# Cofre Digital

pares = 0
impares = 0
soma = 0
numeros = []

for i in range(5):
    while True:
        try:
            numero = int(input(f"Digite o {i+1}º número: "))
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")
    numeros.append(numero)
    soma = soma + numero

    if numero % 2 == 0:
        pares = pares + 1
    else:
        impares = impares + 1

    if i == 0:
        maior = numero
        menor = numero
    else:
        if numero > maior:
            maior = numero
        if numero < menor:
            menor = numero

media = soma / 5

maiores = 0
for i in range(5):
    if numeros[i] > media:
        maiores = maiores + 1

print("Números pares:", pares)
print("Números ímpares:", impares)
print("Maior número:", maior)
print("Soma:", soma)
print("Média:", media)
print("Menor número:", menor)
print("Números maiores que a média:", maiores)
```

```python
# Número espelho

# Solicite um número inteiro positivo

# Inverta o número

# Informe se ele é um palíndromo.


numero = int(input("Digite um número inteiro e    ppositivo: "))

espelho = 0
copia = numero

while copia > 0:
    resto = copia % 10
    espelho = espelho * 10 + resto
    copia = copia // 10

print("Número invertido:", espelho)

if numero == espelho:
    print("É um palíndromo.")
else:
    print("Não é um palíndromo.")
```
