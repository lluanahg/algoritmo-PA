```python
# QUESTÃO 02 – Soma dos Números

numero = int(input("Digite um número inteiro positivo:  "))

soma = 0

for i in range(1, numero + 1):
    soma += i

print("Resultado:", soma)
```

```python
# QUESTÃO 03 – Tabuada

# 6 Tabuada, o usuário digita um número e o programa imprime a tabuada do número

numero = int(input("Digite um número e descubra sua tabuada: "))

for i in range (1, 11):
  print(f"{numero} x {i} = {numero * i}")
```

```python
# QUESTÃO 04 – Fatorial

numero = int(input("Digite um número inteiro e positivo: "))

while numero < 0:
  numero = int(input("Número invalido! Digite um número positivo: "))

fatorial = 1
for i in range(1, numero + 1):
      fatorial *= i

print("Fatotorial = ", fatorial)



```
