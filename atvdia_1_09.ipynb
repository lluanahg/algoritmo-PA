```python
#Exercício 1 — Contagem

#Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

numero = int(input("Digite um número: "))

contador = 1

while contador <= numero:
    print(contador)
    contador += 1
```

```python
#Exercício 2 — Contagem regressiva

#Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

numero = int(input("Digite um número: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

```python
#Exercício 3 — Números pares

#Faça um programa que mostre todos os números pares de 0 até 20 usando while.

numero = 0

while numero <= 20:
   print(numero)
   numero += 2
```

```python
#Exercício 4 — Soma dos números

#Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

numero = int(input("Digite um número: "))

contador = 1
soma = 0

while contador <= numero:
    soma += contador
    contador += 1


print("Soma:", soma)
```

```python
#Exercício 5 — Tabuada

#Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

numero = int(input("Digite um número: "))

contador = 1

while contador <= 10:
    print(numero, "x", contador, "=", numero * contador)
    contador = contador + 1
```

```python
#Exercício 6 — Senha

#Crie um programa que peça uma senha ao usuário.

senha = input("Digite a senha: ")

while senha != "1234":
    print("Senha incorreta!")
    senha = input("Digite a senha: ")

print("Acesso permitido!")
```

```python
#Exercício 7 — Somando até digitar zero

soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número: "))

print("Soma:", soma)
```

```python
#Exercício 8 — Maior número

maior = 0

numero = int(input("Digite um número: "))

while numero != 0:
    if numero > maior:
        maior = numero

    numero = int(input("Digite um número: "))

print("Maior número:", maior)
```

```python
#Exercício 9 — Média das notas

contador = 1
soma = 0

while contador <= 5:
    nota = float(input("Digite a nota: "))
    soma += nota
    contador += 1

media = soma / 5

print("Média:", media)
```

```python
#Exercicio 10

saldo = 1000

opcao = 0

while opcao != 4:
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor = float(input("Digite o valor do depósito: "))
        saldo = saldo + valor
        print("Depósito realizado!")

    elif opcao == 2:
        valor = float(input("Digite o valor do saque: "))

        if valor <= saldo:
            saldo = saldo - valor
            print("Saque realizado!")
        else:
            print("Saldo insuficiente!")

    elif opcao == 3:
        print("Saldo:", saldo)

    elif opcao == 4:
        print("Saindo...")

    else:
        print("Opção inválida!")
```
