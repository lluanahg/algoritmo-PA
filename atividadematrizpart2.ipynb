```python
# exercicio 8-

matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

soma = 0

for linha in matriz:
    soma = soma + linha[1]

print(soma)
```

```python
# Exercicio 9-

matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print(maior)
```
