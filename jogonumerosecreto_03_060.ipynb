```python



# O usuário terá 03 chances para adivinhar um número de 0 a 10
# Se o número secreto for maior ou menor que o palpite o programa irá imprimir uma mensagem informando
# Infomar se o palpite está quente ou frio.

import random
numero_secreto = random.randint(0,10)

tentativa = 1
acertou = False

while tentativa <= 3 and acertou == False:
   try:
      palpite = int(input(f"Tentativa {tentativa}/3 - Digite um número de 0 a 10:"))
   except ValueError:
      print("Entrada inválida. Por favor, digite um número inteiro.")
      continue # Não conta como tentativa se a entrada não for um número

   # Adiciona a validação para números fora do range 0-10
   if not (0 <= palpite <= 10):
      print("Número inválido! Por favor, digite um número entre 0 e 10.")
      continue # Não conta como tentativa se o número estiver fora do range

   if palpite == numero_secreto:
      print("Parabens! Você acertou!")
      acertou = True

   else:
    if palpite < numero_secreto:
       print("O Número secreto é maior")
    elif palpite > numero_secreto:
       print("O Número secreto é menor")

    diferenca = abs(numero_secreto - palpite)
    if diferenca <= 2:
        print("Quente!")
    else:
        print("Frio!")

   tentativa += 1
```
