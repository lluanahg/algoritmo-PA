# ***exercio match***

```python
cor = input("Digite uma cor: ").lower()

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
numero = int(input("Digite um número de 1 a 5: "))

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case 4:
        print("Quatro")
    case 5:
        print("Cinco")
    case _:
        print("Opção inválida")
```

```python
bebida = int(input("Digite uma opção (1, 2 ou 3): "))

match bebida:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
dia = input("Digite um dia da semana: ").lower()

match dia:
    case "sábado" | "sabado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia da semana")

```

```python
nota = input("Digite a nota (A, B ou C): ").upper()

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

# ***EXRCICIO RANGE***


```python
for numero in range(10):
    print(numero)
```

```python
for numero in range(1, 11):
    print(numero)
```

```python
for numero in range(5, 16):
    print(numero)
```

```python
for numero in range(0, 21, 2):
    print(numero)
```
