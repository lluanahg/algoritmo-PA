```python
# Programa para identificar se um número é par ou ímpar

#ENTRADA

numero = int(input("Digite um número: "))

if numero % 2 == 0:
   print("O número é PAR")
else:
   print("O número é ÍMPAR")
```

```python
# Escrever um programa que  o usuario digita o nome e as suas 04 notas referentes aos bimestres, calcula a média e mostra se o aluno foi:
# aprovado, recuperação ou reprovado.
# Nota maior ou igual a 7 aprovado
# Nota maior ou igual a 5 e menor que 7 recuperaçao
# Nota menor que 5 reprovado
# Entrada

nome = str(input("Digite seu nome."))

nota1 = int(input("Digite sua nota do primeiro bimestre: "))

nota2 = int(input("Digite sua nota do segundo bimestre: "))

nota3 = int(input("Digite sua nota do terceiro bimestre: "))

nota4 = int(input("Digite sua nota do quarto bimestre: "))


media = 5

(nota1 + nota2 + nota3 + nota4)/4

if media >= 7:
   print(f"{nome}, Aprovado!")

elif media >= 5:
   print(f"{nome}, Recuperação!")


else:
   print(f"{nome} Reprovado!")
```

```python
# Escrever um programa que recebe os valores de 03 lados e verifica se é um triÂngulo válido
# E classificar o triângulo
# Verificar se o triângulo é válido

# Classisficar tipos de triângulos;
# * Equilater -todos os lados iguais
# * Isoceles -  Possui dois lados iguais
# * Escaleno - todos os lados diferentes

lado1 = int(input("Digite o primeiro lado do triângulo: "))

lado2 = int(input("Digite o segundo lado do triângulo: "))

lado3 = int(input("Digite o terceiro lado do triângulo: "))

if (lado1 + lado2 > lado3) and (lado1 + lado3 > lado2) and (lado2 + lado3 > lado1):
  print("É um triângulo válido")

  if lado1 == lado2 and lado1 == lado3:
    print("É um triângulo equilátero")

  elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print("É um triângulo isósceles")

  else:
    print("É um triângulo escaleno")

else:
  print("Não é um triângulo válido")
```
