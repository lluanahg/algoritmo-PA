```python
# identificar se a palavra  é um palindromo.

palavra = str (input("Digite uma palavra"))








































































































































































































































































































































































































































































invertida = ""

for i in (palavra):

  invertida = i  + invertida
if palavra == invertida:
   print (f"(palavra) É um palíndromo")


else:
   print (f"(palavra) Não é um palíndromo.")
```

```python
# Soma acumulada
# Peça ao usuário para digitar um número e imprimir o resultado da soma 1 até o número digitado

```

```python
numero = int(input("Digite um número inteiro positivo: "))

if numero < 0:
    print("Por favor, digite um número positivo.")
else:
    soma_acumulada = 0
    for i in range(1, numero + 1):
        soma_acumulada += i
    print(f"A soma acumulada de 1 até {numero} é: {soma_acumulada}")
```

```python
# Peça um número ao usuário e oprograma irá separar números pares e impares
# Exemplo de sáida:

# Pares: 0, 2, 4, 6 ...
# Impares: 1,3,5 ...
```

```python
numero = int(input("Digite um número para separar pares e ímpares: "))

pares_str = "Pares: "
impares_str = "Impares: "

primeiro_par = True
primeiro_impar = True

for i in range(numero + 1): # Inclui o 0 e o número digitado
    if i % 2 == 0: # Verifica se é par
        if not primeiro_par:
            pares_str += ", "
        pares_str += str(i)
        primeiro_par = False
    else: # Se não é par, é ímpar
        if not primeiro_impar:
            impares_str += ", "
        impares_str += str(i)
        primeiro_impar = False

print(pares_str)
print(impares_str)
```
