```python
# 4 – Expressões aritméticas
#Calcule e mostre o resultado de:
#• O dobro de 15
#• A metade de 90  #O resto da divisão de 17 por 5
#• 2 elevados à potência 4
#• A média de 10, 8 e 6

dobro = 15 * 2
print("O dobro de 15 é:", dobro)


metade = 90 / 2
print("A metade de 90 é:", metade)

resto = 17 % 5
print("O resto da divisão de 17 por 5 é:", resto)

potencia = 2 ** 4
print("2 elevado à potencia 4 é:", potencia)

media = (10 + 8 + 6) / 3
print("A media de 10, 8 e 6 é:", media)
```
