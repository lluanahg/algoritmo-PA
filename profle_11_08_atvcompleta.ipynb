```python
#Exercício 1 – Idade para dirigir
idade = int(input("Digite sua idade aqui e descubra se poderá tirar sua carteira de motorista: "))

if idade >= 18:
    print("você pode tirar a carteira de motorista.")
else:
    print("Você não pode tirar a carteira de motorista.")
```

```python
#Exercício 2 – Par ou Ímpar. Receba um número e diga se ele é par ou ímpar.

numero = int(input("Digite um número aqui: "))

if numero % 2 == 0:
   print("O número é par! ")

else:
  print("O número é impar!")
```

```python
#Exercício 3 – Maior de dois números. Receba dois números e mostre qual é o maior. Se os dois números forem iguais, diga que eles são iguais.


numero1 = float(input("Digite o primerio número aqui: "))

numero2 = float(input("Digite o segundo número aqui: "))

if numero1 > numero2:
   print("O maior número é:", numero1)

elif numero2 > numero1:

  print("O maior número é", numero2)

else:
  print("Os dois números são iguais.")

```

```python
#exercicio 4 - situação do aluno

print("Digite aqui sua nota e descubra se passou de ano!")

nota = int(input("Coloque aqui sua nota: "))

if nota >= 7:
  print("Aprovado!")

elif nota >= 5 < 7:
  print("Recuperação")

else:
  print("Reprovado!")





```

```python
#Exercício 5 – Desconto na compra

valor = float(input("Digite o valor da compra: "))

if valor >= 100:
  desconto = valor * 0.10
  valor_final = valor - desconto
  print("Valor final:", valor_final)

else:
  print("Valor normal da compra:", valor)
```

```python
#Exercício 6 – Positivo, Negativo ou Zero

numero = float(input("Digite um número aqui e descubra se ele é positivo, negativo ou igual a zero: "))

if numero > 0:
  print("O número é positivo!")
elif numero < 0:
  print("O número é negativo!")

else:
  print("O número digitado é igual a zero.")
```

```python
#Exercício 7 – Classificação de idade

idade = int(input("Digite sua idade aqui: "))

if idade < 12:
  print("Criança.")

elif idade <= 17:
   print("Adolescente")

elif idade <= 59:
   print("Adulto")

else:
  print("Idoso")
```

```python
# Exercício 8 – Média de duas notas

nota1 = float(input("Digite a primeira nota aqui: "))

nota2 = float(input("Digite a segunda nota aqui: "))

media = (nota1 + nota2) / 2

if media >= 7:
  print("Aprovado")

else:
  print("Reprovado")
```

```python
# Exercicio 10- calculadora de IMC

peso = float(input("Digite seu peso  em kg aqui e descubra seu imc: "))

altura = float(input("Agora digite sua altura em metros: "))

imc = peso/ (altura * altura)

print("Seu IMC é:", imc)

if imc < 18.5:
  print("Você está abaixo do peso.")

elif imc < 25:
   print("Você está em um peso ideal.")

elif imc < 30:
  print("Você pode estar sobrepeso.")

else:

 print("Você pode estar obeso!")
```
